Boundaries catch more bugs because they sit right on the edge of a rule.A middle value (like 85) is safely inside the correct grade, so even a small mistake in the code still gives the right answer.But a boundary value (like 70) is standing exactly where the rule changes.

## Lab 10. The accumulator pattern

Three steps, every single time:

1. **initialise** a variable before the loop
2. **update** it inside the loop
3. **use** it after the loop

````python
theSum = 0                 # 1. initialise

for count in range(1, 6):
    theSum += count        # 2. update
    print("so far:", theSum)

print("Final total:", theSum)   # 3. use
````
