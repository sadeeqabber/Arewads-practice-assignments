````python
text = "the cat and the hat and the bat"
counts = {}
for word in text.split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
````

### TO DO 2 (Level 2: Apply): count_even

Write `count_even(numbers)` that returns how many even numbers are in the list. It must return 0 for an empty list.

````python
# TO DO 2
def count_even(numbers):
    """Return how many even numbers are in the list."""
    count = 0
    for number in numbers:
        if number % 2 == 0:
            count += 1
    return count
````

````python
# CHECK: run this cell. Do not change it.
assert count_even([1, 2, 3, 4, 6]) == 3
assert count_even([]) == 0, "an empty list has 0 even numbers"
assert count_even([7, 9]) == 0
assert count_even([-2, 0]) == 2, "-2 and 0 are even"
print("All checks passed")
````

### TO DO 3 (Level 2: Apply): above_average

Write `above_average(numbers)` that **returns a new list** with the values greater than the average. It must not change the list passed in.

````python
# TO DO 3
def above_average(numbers):
    """Return a NEW list of the values greater than the average."""
    if not numbers:
        return []
    avg = sum(numbers) / len(numbers)
    result = []
    for number in numbers:
        if number > avg:
            result.append(number)
    return result
````

````python
# CHECK: run this cell. Do not change it.
data = [70, 85, 90, 55, 100]
assert above_average(data) == [85, 90, 100], "average is 80"
assert data == [70, 85, 90, 55, 100], "do not change the original list"
assert above_average([5, 5, 5]) == [], "nothing is above the average"
print("All checks passed")
````

### TO DO 4 (Level 2: Apply): tally

Write `tally(items)` that takes a list of strings and returns a dictionary counting how many times each one appears.

````python
# TO DO 4
def tally(items):
    """Return a dictionary counting how often each item appears."""
    counts = {}
    for item in items:
        counts[item] = counts.get(item, 0) + 1
    return counts
````

````python
# CHECK: run this cell. Do not change it.
assert tally(["yes", "no", "yes", "yes"]) == {"yes": 3, "no": 1}
assert tally([]) == {}
assert tally(["a"]) == {"a": 1}
print("All checks passed")
````

### TO DO 5 (Level 3: Extend): Averages per student from a file

Write `student_averages(filename)` that reads lines like `Ana,88` and returns a dictionary from each name to that student's **average** score, rounded to 1 decimal. Hint: first build a dictionary from name to a **list** of scores.

````python
# TO DO 5
def student_averages(filename):
    """Read lines like 'Ana,88' and return {name: average score (1 decimal)}."""
    scores_by_name = {}
    with open(filename) as f:
            for line in f:
                line = line.strip()              # remove
                if line == "":
                    continue
                name, score_text = line.split(",")
                score = int(score_text)
                if name not in scores_by_name:
                    scores_by_name[name] = []
                scores_by_name[name].append(score)

    averages = {}
    for name, scores in scores_by_name.items():
        averages[name] = round(sum(scores) / len(scores), 1)
    return averages
````

````python
# CHECK: run this cell. Do not change it.
result = student_averages("grades.txt")
assert result == {"Ana": 87.0, "Ben": 76.0, "Chidi": 81.0}, f"got {result}"
print("All checks passed")
````

**Question 1.** After `a = [1, 2, 3]` and `b = a` and `b.append(4)`, what is `a`? Explain, and show one way to get an independent copy.

_Your answer:_  a is [1, 2, 3, 4].

b = a does not make a copy. It gives the same list a second name, so a and b point to one list in memory. b.append(4) changes that one list in place, and since a is looking at the same list, a shows the 4 too (this is called aliasing).

To get an independent copy, use any of these:

````python
# Aliasing: two names, one list
a = [1, 2, 3]
b = a
b.append(4)
print("a =", a, "  same list?", a is b)

# An independent copy
a = [1, 2, 3]
b = a.copy()          # or a[:]  or list(a)
b.append(4)
print("a =", a, " b =", b, "  same list?", a is b)
````

**Question 3.** `scores = scores.sort()` leaves `scores` as None. Why? Give two correct alternatives.

_Your answer:_ 
scores.sort()` sorts the list **in place**: it rearranges the existing list and **returns `None`** (nothing). Then `scores = scores.sort()` stores that `None` into `scores`, replacing the list we just sorted, so the list is lost.

Two correct alternatives:

1. Call it on its own line and do not assign the result: `scores.sort()`
2. Use `sorted()`, which **returns a new sorted list**: `scores = sorted(scores)`, or keep the original and make a new name: `ordered = sorted(scores)`


````python
# Two correct alternatives for sorting a list
scores = [88, 92, 79]

scores.sort()                  # 1) sort in place, do NOT assign the result
print(scores)                  # [79, 88, 92]

scores = [88, 92, 79]
scores = sorted(scores)        # 2) sorted() returns a NEW sorted list, so assigning is fine
print(scores)                  # [79, 88, 92]
````
