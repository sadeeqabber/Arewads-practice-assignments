### The most important rule today: indentation

In many languages, indentation is only for beauty. **In Python, indentation is grammar.** It tells Python which lines belong *inside* an `if` or a loop.

```python
if temperature > 35:
    print("It is hot.")         # indented: INSIDE the if, runs only when the condition is True
    print("Drink water.")       # indented: also inside
print("Goodbye.")               # not indented: OUTSIDE, always runs
```

Three rules:

1. The line that opens a block (`if`, `else`, `elif`, `for`, `while`) **ends with a colon `:`**.
2. The lines inside the block are indented by **4 spaces** (press `Tab` once in Jupyter, VS Code, or Colab).
3. The block ends when the indentation goes back.

Two new errors you will meet today:

| Error | Plain meaning | Typical cause |
|---|---|---|
| `SyntaxError: expected ':'` | "You opened a block but forgot the colon." | `if x > 5` |
| `IndentationError` | "I expected indented lines here", or "these lines do not line up." | Forgetting to indent after a colon, or mixing 3 and 4 spaces |

Before a program can choose, it has to ask a yes or no question. In Python the answer is `True` or `False`.

### Read this first

Python has a type with only two values: **`True`** and **`False`** (capital first letter, no quotes). The type is called `bool`, after the mathematician George Boole.

You get a `bool` whenever you **compare** two values:

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | equal to | `5 == 5` | `True` |
| `!=` | not equal to | `5 != 5` | `False` |
| `<` | less than | `3 < 5` | `True` |
| `>` | greater than | `3 > 5` | `False` |
| `<=` | less than or equal to | `5 <= 5` | `True` |
| `>=` | greater than or equal to | `4 >= 5` | `False` |

**Warning, the number one beginner mistake:**

- `=` (one sign) **stores** a value: `age = 18` means "put 18 in `age`". It is an instruction.
- `==` (two signs) **asks** a question: `age == 18` means "is `age` equal to 18?". The answer is `True` or `False`.

````python
# WORKED EXAMPLE 1: asking yes/no questions.

price = 1500
budget = 2000

print(price < budget)        # is 1500 less than 2000?            -> True
print(price == budget)       # is 1500 equal to 2000?             -> False
print(price != budget)       # is 1500 different from 2000?       -> True
print(price <= 1500)         # is 1500 less than OR EQUAL to 1500 -> True (the "equal" part makes it True)
print(price < 1500)          # is 1500 strictly less than 1500?   -> False
````

````python
# The answer to a comparison is a value like any other. You can store it in a variable.
canAfford = price <= budget
print(canAfford)
print(type(canAfford))  # <class 'bool'>
````

````python
# Variable names that start with "is", "can", "has" read nicely for booleans.
passMark = 50
score = 47
hasPassed = score >= passMark
print("Has the student passed?", hasPassed)
````

## Lab 2. The `if` statement, one-way selection

The simplest decision: do something, or do nothing.

### Goals

### Read this first

You make this kind of decision every day: *"If it is raining, I take an umbrella."* If it is not raining, you do nothing special. You just carry on.

```
            |
            v
      +-----------+   False
      | condition |-----------+
      +-----------+           |
            | True            |
            v                 |
      indented lines          |
            |                 |
            +<----------------+
            v
      the program continues
```

The shape in Python:

```python
if condition:          # 1. the word if   2. a condition   3. a colon
    statement          # 4. indented lines: run ONLY when the condition is True
    statement
next statement         # not indented: runs ALWAYS
```

````python
# WORKED EXAMPLE 2: do something, or do nothing.

balance = 800
price = 1500

# The indented lines run only when the condition is True.
if price > balance:
    print("Insufficient balance.")                     # inside the if
    print("You need", price - balance, "naira more.")  # inside the if (same indentation)

print("Thank you for using our service.")              # outside the if: ALWAYS runs

# Now the same code with a different balance. The condition is False, so the block is skipped.
balance = 5000
if price > balance:
    print("Insufficient balance.")

print("Thank you for using our service.")
````

## Lab 3. `if-else`, two-way selection

### Goals
By the end of this lab you can:

### Read this first

A one-way `if` does something or nothing. Often you want **this OR that**: *"If it is raining I take a keke, otherwise I walk."*

```
                  |
                  v
   True     +-----------+     False
   +--------| condition |--------+
   |        +-----------+        |
   v                             v
 if-block                   else-block
   |                             |
   +-------------+---------------+
                 v
       the program continues
```

```python
if condition:
    block A          # runs when the condition is True
else:                # else has NO condition, but it has a colon
    block B          # runs when the condition is False
```

**Exactly one** of the two blocks runs. Never both. Never neither. `else` lines up with its `if`.

````python
# WORKED EXAMPLE 3: this, or that.

score = 47
PASS_MARK = 50

if score >= PASS_MARK:
    print("Congratulations, you passed.")
else:
    print("You did not pass this time.")
    print("You were", PASS_MARK - score, "marks short.")     # the else-block can have several lines too

print("--- end of report ---")                               # outside: always runs

# A classic use: protect a calculation that is not always possible.
totalMarks = 0
numberOfTests = 0

if numberOfTests > 0:
    print("Average:", totalMarks / numberOfTests)
else:
    print("No tests yet, so there is no average.")           # without this check: ZeroDivisionError
````

## Lab 4. Multiway selection with `elif`

### Goals
By the end of this lab you can:

### Read this first

`elif` is short for **"else if"**. It lets you ask a second question only when the first answer was "no".

```python
if condition1:
    block 1
elif condition2:      # asked ONLY when condition1 was False
    block 2
elif condition3:      # asked ONLY when condition1 AND condition2 were False
    block 3
else:                 # optional: when every condition above was False
    block 4
```

Python reads the chain **from top to bottom** and runs the block of the **first** condition that is `True`. Then it **jumps to the end of the chain**. The remaining conditions are not even looked at.

Think of a market trader grading mangoes by size. She asks, in this order: *"Is it extra large? No. Is it large? No. Is it medium? Yes."* and drops it in the medium basket. She **stops asking** at the first "yes". She also knows, without checking again, that a mango that reaches the "medium" question is not large, because it already failed that question.

Now imagine she asks the easiest question first: *"Is it bigger than a lime?"* Every mango says yes, and every mango lands in the first basket. That is exactly the bug in 4.2 below.

````python
# WORKED EXAMPLE 4: one choice out of many.

temperature = 31

if temperature >= 40:
    advice = "Extreme heat. Stay indoors."
elif temperature >= 33:            # we only get here when temperature < 40
    advice = "Very hot. Drink plenty of water."
elif temperature >= 25:            # we only get here when temperature < 33
    advice = "Warm. A normal day."
elif temperature >= 15:            # we only get here when temperature < 25
    advice = "Cool. Harmattan is here, take a sweater."
else:                              # everything below 15
    advice = "Cold!"

print(temperature, "degrees:", advice)

# Notice we never wrote  "temperature >= 25 and temperature < 33".
# The "< 33" part is GUARANTEED, because the branch above already caught everything >= 33.
````

## Lab 5. Logical operators: `and`, `or`, `not`

### Goals
By the end of this lab you can:
- combine conditions with `and`, `or`, and `not`
- fill in a truth table
- write a range check such as "between 0 and 100"

### Read this first

Sometimes one question is not enough. *"You may enter the exam hall if you have your ID card **and** you have paid your fees."*

| Operator | Plain meaning | `True` when ... | Everyday example |
|---|---|---|---|
| `p and q` | both | **both** are True | ID card **and** fees paid |
| `p or q` | at least one | **at least one** is True | pay by cash **or** by transfer |
| `not p` | the opposite | `p` is False | **not** absent |

Two things to know:

- `and` is **strict**: one `False` spoils everything.
- `or` is **generous**: one `True` is enough. Note that Python's `or` also gives `True` when **both** are true.

**The range check.** "Is `x` between 0 and 100?" is really two questions: *is it at least 0* **and** *is it at most 100*:

```python
x >= 0 and x <= 100
```

Each side of `and` / `or` must be a **complete** comparison. You cannot write `x >= 0 and <= 100`.

````python
# WORKED EXAMPLE 5: combining conditions.

hasIdCard = True
hasPaidFees = False

# and: BOTH must be True
print("May enter the hall:", hasIdCard and hasPaidFees)     # False, the fees are missing

# or: ONE is enough
hasCash = False
hasTransferApp = True
print("Can pay:", hasCash or hasTransferApp)                # True

# not: flips True <-> False
isAbsent = False
print("Is present:", not isAbsent)                          # True

# The range check: is the age between 18 and 35, both included?
age = 27
isYouth = age >= 18 and age <= 35
print("Qualifies for the youth programme:", isYouth)

# The opposite question: is the age OUTSIDE that range?  Two ways, same answer.
print(age < 18 or age > 35)            # outside means: too low OR too high
print(not (age >= 18 and age <= 35))   # or simply: NOT inside

# A compound condition inside an if:
score = 82
attendance = 60
if score >= 50 and attendance >= 75:
    print("Certificate awarded.")
else:
    print("No certificate: you need a pass mark AND 75 percent attendance.")
````

## Lab 6. Short-circuit evaluation

### Goals
By the end of this lab you can:
- explain what short-circuit evaluation is
- use it to protect a risky operation, such as a division

### Read this first

Python is lazy in a clever way. It evaluates `and` / `or` **from left to right and stops as soon as the answer is certain**.

| Expression | If the left side is ... | ... then Python | Why |
|---|---|---|---|
| `p and q` | `False` | **does not look at `q`** | `False and anything` is `False` |
| `p or q` | `True` | **does not look at `q`** | `True or anything` is `True` |

Everyday version: *"You may enter if you have an ID card and you have paid."* The guard sees you have no ID card. He does not bother checking your receipt. The answer is already no.

Why this matters: you can put a **safety check on the left** and the **risky operation on the right**. If the safety check fails, the risky part is never run.

````python
# WORKED EXAMPLE 6: the guard on the left protects the risky part on the right.

students = 0
totalMarks = 0

# Risky:  totalMarks / students   crashes with ZeroDivisionError when students is 0.
# Safe:   ask "students > 0" FIRST. When it is False, Python never reaches the division.
if students > 0 and totalMarks / students >= 50:
    print("The class average is a pass.")
else:
    print("No students, or the average is below 50.")

# Proof that the right side is really skipped. 1 / 0 would crash, but it is never evaluated:
print(False and 1 / 0 > 1)      # False   (stopped after the left side)
print(True or 1 / 0 > 1)        # True    (stopped after the left side)

# ORDER MATTERS. With the risky part on the LEFT, it would run first and crash:
#   if totalMarks / students >= 50 and students > 0:     <-- ZeroDivisionError
````

## Lab 8. The `for` loop, the basics

### Read this first

A teacher calling the register is running a `for` loop: *"**for each** name **in** the register: call the name, mark present or absent."*

```python
for variable in sequence:     # the HEADER: ends with a colon
```

- Each run of the body is called a **pass** or an **iteration**.

**`range(n)`** produces the `n` whole numbers from `0` to `n - 1`. `range(5)` is `0, 1, 2, 3, 4`. That is **five** numbers, starting at **zero**, and **5 itself is not included**. Programmers count from zero. You will get used to it.

#### Tracing a loop

The best way to understand a loop is to play computer and fill in a **trace table**, one row per pass:

```python
for count in range(3):

```

| pass | `count` | what is printed |
|---|---|---|
| 1st | 0 | 0 |
| 2nd | 1 | 10 |
| 3rd | 2 | 20 |
| | (the range is used up, the loop ends) | |

````python
# WORKED EXAMPLE 8: three ways of using a for loop.

# 1) Simply repeat something 4 times. We do not use the loop variable at all.
for eachPass in range(4):
    print("Ina kwana!")

print("-" * 20)

# 2) Use the loop variable. range(4) gives 0, 1, 2, 3.
for number in range(4):
    print("number is", number)

print("-" * 20)

# 3) People count from 1. If you want 1, 2, 3, 4, add 1 when you USE the variable ...
for number in range(4):
    print("Student", number + 1)

print("-" * 20)

# 4) Loop directly over a list of values. No range, no numbers, just "for each item".
for city in ["Kano", "Kaduna", "Sokoto", "Maiduguri"]:
    print("Sannu,", city + "!")

print("The loop is over.")       # not indented, so it runs once, at the end
````

## Lab 10. The accumulator pattern

Three steps, every single time:

1. **initialise** a variable before the loop
2. **update** it inside the loop
3. **use** it after the loop

### Goals
By the end of this lab you can:
- add up, count, or multiply values with a loop
- choose the correct starting value for an accumulator
- trace an accumulator by hand

### Read this first

How do you add up a long receipt on a calculator? You start at **0**. You add the first price. You add the second price **to what you already have**. You continue. At the end, the display shows the total. The calculator display is an **accumulator**: a variable that collects a result little by little.

This is the most important loop pattern in programming. It always has the same three steps:

```python
total = 0                    # 1. INITIALISE, before the loop
for price in prices:
    total += price           # 2. UPDATE, inside the loop
print(total)                 # 3. USE, after the loop
```

#### Trace table for `range(1, 6)` (the numbers 1 to 5)

| pass | `count` | `theSum` before | `theSum += count` | `theSum` after |
|---|---|---|---|---|
| (start) | | | | 0 |
| 1st | 1 | 0 | 0 + 1 | 1 |
| 2nd | 2 | 1 | 1 + 2 | 3 |
| 3rd | 3 | 3 | 3 + 3 | 6 |
| 4th | 4 | 6 | 6 + 4 | 10 |
| 5th | 5 | 10 | 10 + 5 | **15** |

Five flavours of the same pattern:

| You want to ... | Start with | Update with |
|---|---|---|
| **add up** values | `0` | `total += value` |
| **count** how many | `0` | `counter += 1` |
| **multiply** values | `1` | `product *= value` |
| **build** a string | `""` | `text += piece` |
| **build** a list | `[]` | `items.append(value)` |

````python
# WORKED EXAMPLE 10: three accumulators in one loop.

prices = [1500, 700, 2300, 450, 1200]      # a shopping basket

total = 0              # 1. initialise: a SUM starts at 0
itemCount = 0          # 1. initialise: a COUNTER starts at 0
expensiveCount = 0     # 1. initialise: another counter

for price in prices:
    total += price                 # 2. update: add this price to the running total
    itemCount += 1                 # 2. update: one more item seen
    if price > 1000:               #    an if inside a loop: only SOME items are counted
        expensiveCount += 1
    print("after", price, "-> total so far", total)

# 3. use the results, AFTER the loop (not indented)
print("Items:", itemCount)
print("Total:", total)
print("Average price:", total / itemCount)
print("Items above 1000 naira:", expensiveCount)
````

### Read this first

| Form | Meaning | Example | Produces |
|---|---|---|---|
| `range(stop)` | from 0 up to, **not including**, `stop` | `range(4)` | 0, 1, 2, 3 |
| `range(start, stop)` | from `start` up to, **not including**, `stop` | `range(1, 5)` | 1, 2, 3, 4 |
| `range(start, stop, step)` | the same, jumping by `step` | `range(1, 10, 3)` | 1, 4, 7 |
| negative `step` | count **down**. `start` must be the bigger one. | `range(5, 0, -1)` | 5, 4, 3, 2, 1 |

**The golden rule: the `stop` value is never included.** To go from 1 **through** 10, write `range(1, 11)`. In general, "from `a` through `b`" is `range(a, b + 1)`.

A `range` does not show its numbers when printed. Wrap it in `list(...)` to see them. This is a good way to check a range before you put it in a loop.

````python
# WORKED EXAMPLE 11: start, stop, step.

print(list(range(6)))             # [0, 1, 2, 3, 4, 5]     one argument: the stop
print(list(range(1, 7)))          # [1, 2, 3, 4, 5, 6]     two arguments: start, stop
print(list(range(0, 21, 5)))      # [0, 5, 10, 15, 20]     three: start, stop, step. 21 so that 20 is included
print(list(range(3, 0, -1)))      # [3, 2, 1]              counting down. 0 is the stop, so it is NOT included

# A countdown
for seconds in range(3, 0, -1):
    print(seconds, "...")
print("Go!")

# The years of a four-year degree that starts in 2026
for year in range(2026, 2030):
    print("Academic year", year)

# Every second week of a 12-week course: 2, 4, ..., 12
for week in range(2, 13, 2):
    print("Quiz in week", week)
````

## Lab 12. Off-by-one errors

An off-by-one error is not a syntax error. Python runs happily and gives you a wrong answer.

### Read this first

An **off-by-one error** means your loop runs one time too many or one time too few. It is the most common loop bug in the world, and experienced programmers still make it.

Why is it so easy? Try this: *you build a straight fence 10 metres long with a post every metre. How many posts do you need?* Most people say 10. The answer is **11**. Counting things and counting the gaps between things are different.

With `range`, the usual cause is forgetting that **`stop` is not included**.

**How to catch it:** check the **first** and the **last** value your loop produces. Do not check the middle. The middle is always fine.

| 1 through 4 | `range(1, 4)` → 1, 2, 3 | `range(1, 5)` |
| 1 through `n` | `range(n)` → 0 ... n-1 | `range(1, n + 1)` |
| `n` items, do not care which numbers | | `range(n)` |

## Lab 13. Traversing a string

A string is a sequence of characters, so a `for` loop can walk through it directly.

### Goals
By the end of this lab you can:
- visit every character of a string with a `for` loop
- build a new string with the accumulator pattern

### Read this first

A `for` loop works on **any sequence**. A list is a sequence of items. A `range` is a sequence of numbers. A **string is a sequence of characters**. So:

```python
for character in "Kano":
    ...
```

runs the body four times, with `character` equal to `"K"`, then `"a"`, then `"n"`, then `"o"`.

**New detail: `end=`.** Normally `print` finishes with a new line. `print(x, end=" ")` finishes with a space instead, so the next `print` continues on the **same** line.

**Building a string** is the accumulator pattern again. Start with the empty string `""`, then add one piece per pass.

| pass | `character` | `result += character.upper()` | `result` after |
|---|---|---|---|
| (start) | | | `""` |
| 1st | `"k"` | `"" + "K"` | `"K"` |
| 2nd | `"a"` | `"K" + "A"` | `"KA"` |
| 3rd | `"n"` | `"KA" + "N"` | `"KAN"` |

````python
# WORKED EXAMPLE 13: walking through a string.

city = "Sokoto"

# 1) Visit each character.
for character in city:
    print(character, end="-")        # end="-" keeps everything on one line
print()                              # an empty print() just ends the line

# 2) Count something: how many times does the letter o appear?
oCount = 0
for character in city:
    if character == "o":
        oCount += 1
print("Number of o's:", oCount)

# 3) Build a NEW string, one character at a time (string accumulator, starts at "").
spaced = ""
for character in city:
    spaced += character + " "        # new piece goes on the RIGHT: order is kept
print(spaced)

# 4) Same idea, but the new piece goes on the LEFT. What does that do to the order?
mystery = ""
for character in "abc":
    mystery = character + mystery
    print("so far:", mystery)
````

## Lab 14. Loops and decisions together

This is where Part 1 and Part 2 meet. The `if` goes inside the loop body.

### Goals
By the end of this lab you can:

- keep track of two levels of indentation

### Read this first

A loop looks at **every** item. An `if` inside the loop lets you treat **some** items differently. Together they give you the two most useful patterns in data science:

- **filter**: keep (or count) only the items that pass a test
- **classify**: put each item into a category

Watch the indentation. There are now **two levels**:

```python
for score in scores:            # level 0
    if score >= 50:             # level 1: inside the for        (4 spaces)
        passCount += 1          # level 2: inside the if, inside the for   (8 spaces)
print(passCount)                # level 0: after the loop
```

You met the **`in`** operator in the loop header. On its own it is also a question: `"a" in "Kano"` asks *"does the letter a appear in Kano?"* and gives `True`.

````python
# WORKED EXAMPLE 14: filter and classify.

temperatures = [29, 36, 41, 33, 38, 27, 40]     # one reading per day, in degrees

hotDays = 0                    # counter
hotList = []                   # list accumulator

for temp in temperatures:
    if temp >= 38:                         # FILTER: only some days pass this test
        hotDays += 1
        hotList.append(temp)
        print(temp, "-> hot")
    elif temp >= 30:                       # CLASSIFY: every day gets a label
        print(temp, "-> warm")
    else:
        print(temp, "-> mild")

print("Hot days:", hotDays)
print("Their temperatures:", hotList)

# The "in" operator as a question:
print("a" in "Kano")           # True
print("z" in "Kano")           # False

# Used as a filter: count the digits in a phone number written with spaces and dashes.
phone = "0803-123 4567"
digitCount = 0
for character in phone:
    if character in "0123456789":
        digitCount += 1
print("Digits:", digitCount)
````

## Lab 15. Formatting text for output

### Goals
By the end of this lab you can:
- print numbers in columns that line up
- control the number of decimal places

### Read this first

Last week, `str(4500.0)` gave you `4500.0` and you could do nothing about it. Now you can. A **format string** is a template with **holes**. Each hole says how wide the value should be and how it should look.

The slides use the `%` operator:

```python
"%-10s%6d%10.2f" % ("Rice", 3, 4500)
```

| Hole | Meaning |
|---|---|
| `%d` | a whole number (**d**ecimal integer) |
| `%f` | a **f**loat |
| `%s` | a **s**tring |
| `%6d` | an integer in a field 6 characters wide, **right**-aligned |
| `%-6d` | the same, **left**-aligned (the minus sign means left) |
| `%10.2f` | a float, 10 characters wide in total, with exactly **2** decimal places |
| `%0.2f` | a float with 2 decimal places, as narrow as possible |

After the string comes `%` and then the values. One value may stand alone. Several values go in **parentheses**, in the same order as the holes.

Rule of thumb: **text left-aligned, numbers right-aligned**, as in any printed price list.

````python
# WORKED EXAMPLE 15: holes in a template.

# 1) Width and alignment. The | characters let you SEE where each field starts and ends.
print("|%5d|" % 42)           # |   42|   width 5, right-aligned (the default)
print("|%-5d|" % 42)          # |42   |   width 5, left-aligned
print("|%8s|" % "Kano")       # |    Kano|
print("|%-8s|" % "Kano")      # |Kano    |

# 2) Decimal places. Note that the value is ROUNDED, not cut.
print("%0.2f" % 3.14159)      # 3.14
print("%0.2f" % 2.5)          # 2.50   always two decimals, good for money
print("%0.0f" % 2.71828)      # 3

# 3) Several holes: the values go in parentheses, in order.
print("%s is %d years old" % ("Aisha", 21))

# 4) A small table. Text left, numbers right.
print("%-10s%8s" % ("State", "Temp"))
print("%-10s%8.1f" % ("Kano", 38.25))
print("%-10s%8.1f" % ("Jos", 24.0))
print("%-10s%8.1f" % ("Maiduguri", 41.5))
````

## Lab 16. The `while` loop

`for` repeats a known number of times. `while` repeats until a condition changes, which you may not be able to predict in advance.

### Goals
By the end of this lab you can:
- write a `while` loop with its three essential parts
- explain what causes an infinite loop, and how to stop one
- choose between `for` and `while`

### Read this first

A `for` loop is for when you know **how many times** (or over which items) to repeat. Sometimes you do not know. *"Keep stirring the tuwo **until** it is thick."* *"Keep asking for the PIN **until** it is correct."* How many times? Nobody knows in advance. That is a `while` loop.

```python
while condition:      # "as long as the condition is True, keep going"
    body
```

Python checks the condition. If it is `True`, the body runs, and Python goes **back up** and checks again. When the condition is `False`, the loop ends. The condition is only checked at the **top**, before each pass.

```
            |
            v
      +-----------+   False
  +-->| condition |-----------> after the loop
  |   +-----------+
  |         | True
  |         v
  |       body
  +---------+
```

**Every `while` loop needs three parts.** Forget any one and it breaks:

```python
count = 1                 # 1. INITIALISE the loop control variable
while count <= 5:         # 2. TEST it
    print(count)
    count += 1            # 3. UPDATE it, so that the test will one day be False
```

If part 3 is missing, the condition stays `True` for ever: an **infinite loop**. The cell shows `[*]` and never finishes. Press the **stop button** (the black square).

| | `for` | `while` |
|---|---|---|
| Use when | you know how many times, or you walk over a sequence | you repeat until something happens |
| Also called | **definite** iteration | **indefinite** iteration |
| Who updates the loop variable? | Python does it for you | **You** must do it |
| Risk of an infinite loop | none | yes |

````python
# WORKED EXAMPLE 16: repeat until something changes.

# Saving up for a phone. We do NOT know in advance how many months it takes.
PHONE_PRICE = 95000
MONTHLY_SAVING = 12000

savings = 0                          # 1. initialise
months = 0

while savings < PHONE_PRICE:         # 2. test: keep going as long as we do not have enough
    savings += MONTHLY_SAVING        # 3. update: this is what will make the condition False one day
    months += 1
    print("Month", months, "- saved so far:", savings)

print("You can buy the phone after", months, "months.")
print("Money left over:", savings - PHONE_PRICE)

# Trace it by hand for the first three passes:
#   pass | savings before | test 'savings < 95000' | savings after | months
#   1st  |       0        |        True            |     12000     |   1
#   2nd  |   12000        |        True            |     24000     |   2
#   3rd  |   24000        |        True            |     36000     |   3
#   ...
#   8th  |   84000        |        True            |     96000     |   8
#   9th? |   96000        |        False  -> the loop ends
````

## Lab 17. `while True` and `break`

Notice the duplicated `input` line in Lab 16.4, once before the loop and once at the bottom of the body. `while True` with `break` removes it.

### Goals
By the end of this lab you can:
- leave a loop from the middle with `break`
- write an input-validation loop

### Read this first

`while True:` looks like a guaranteed infinite loop, because the condition can never become `False`. It is, **unless** the body contains a **`break`**. `break` means: *leave the loop immediately and continue with the first line after it.*

```python
while True:
    get a value
    if the value is good:
        break              # jump out of the loop
    complain               # only reached when the value was bad
# we arrive here after the break
```

This is the natural shape for **"keep asking until the answer is acceptable"**: the question appears only once in the code, and the exit is in the middle, right where the decision is made.

`break` always belongs inside an `if`. A `break` without an `if` would stop the loop on the very first pass.

````python
# WORKED EXAMPLE 17: keep asking until the answer is acceptable.
# We simulate a user who types three wrong PINs and then the right one,
# so that this cell runs without pausing. In a real program:  attempt = input("Enter your PIN: ")

CORRECT_PIN = "4321"
typedByUser = ["1234", "0000", "4312", "4321", "9999"]     # pretend keyboard input
position = 0

while True:
    attempt = typedByUser[position]        # real program: attempt = input("Enter your PIN: ")
    position += 1
    print("User typed:", attempt)

    if attempt == CORRECT_PIN:
        print("PIN accepted.")
        break                              # leave the loop NOW. The lines below are skipped.

    print("Wrong PIN, try again.")         # only reached when there was no break

print("Welcome! It took you", position, "attempts.")
# Notice: "9999" was never read. The loop had already ended.
````

Games, simulations, and random sampling in data science all need numbers that are **not predictable**. They come from the `random` module:

| Function | Gives you |

## Part 2 summary: which loop?

| Situation | Loop |
|---|---|
| Repeat exactly `n` times | `for i in range(n):` |
| Go from `a` through `b` | `for i in range(a, b + 1):` |

Loop over the scores once and compute:

Do not use `max`, `min`, `sum` or `len` inside your loop. Use comparisons and accumulators.

<details>
<summary><b>Hint 1</b> (click to open, try first!)</summary>

How do you find the tallest person in a queue without seeing everyone at once? Remember the tallest **so far**; compare each new person with that. Start `highest` and `lowest` with the **first score**, `scores[0]`.

</details>

<details>
<summary><b>Hint 2</b> (click to open, try first!)</summary>

Inside the loop: `if score > highest: highest = score`, the same with `<` for `lowest`, `total += score`, and count the passes. The average is computed **after** the loop: `total / 10`, or keep your own counter.

</details>

````python
scores = [72, 45, 88, 50, 91, 33, 67, 49, 100, 58]

highest = scores[0]
lowest = scores[0]
total = 0
passCount = 0

# TODO

for score in scores:
    if score > highest:
        highest = score
    if score < lowest:
        lowest = score
    total += score
    if score >= 50:
        passCount += 1

average = total / len(scores)

print(highest, lowest, average, passCount)
````
