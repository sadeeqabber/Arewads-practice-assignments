## A.1 Defining and calling functions

### `def` defines a function; `name()` calls it

```python
def name():
    body      # indented, like the body of an if or a loop
```

- `def name():` followed by an indented body **defines** the function.
- Writing `show_rule()` **calls** it, which runs the body.
- You can call it as many times as you like.

This is the same shape as the built-ins you already use: a name followed by parentheses.

### ▶️ TRY IT

````python
def show_rule():
    print("-" * 20)

show_rule()
print("Report")
show_rule()
````

### Parameters: send information in

A function becomes much more useful when the caller can send it information.

- `message` is a **parameter**: a variable that lives inside the function.
- `"Welcome"` is the **argument**: the value passed in the call.
- Each call gets its own value.

### ▶️ TRY IT

````python
def banner(message):
    print("*" * 12)
    print("*", message)
    print("*" * 12)

banner("Welcome")
banner("Goodbye")
````

Compare this with the warm-up: same output, but the pattern is written **once** and has a **name**.

### Several parameters: matched by position

- The first argument goes to the first parameter, the second to the second, and so on.
- The last line below swaps them, and Python cannot tell!
- **Order matters.**

````python
def describe(name, age):
    print(f"{name} is {age} years old")

describe("Ana", 19)
describe(20, "Ben")
````

## A.2 Returning a value

### `return` sends a value back to the caller

- `return value` **ends** the function and **gives back** the value.
- The call `square(4)` *becomes* `16`.
- You can use the result in assignments, expressions, or `print`.

### ▶️ TRY IT

````python
def square(x):
    return x * x

answer = square(4)
print(answer)
print(square(3) + square(4))
````

### Return the result; let the caller decide what to do

Compare the two versions side by side:

| `print` inside: the result is lost | `return`: the result can be used |
|---|---|
| `def add_print(a, b):`<br>`    print(a + b)` | `def add_return(a, b):`<br>`    return a + b` |
| Output: `5` then `total is None` | Output: `total is 5` |

````python
def add_return(a, b):
    return a + b

total = add_return(2, 3)
print("total is", total)
````

> **Rule of thumb:** functions that *compute* something should **return** it. Printing is the caller's job.

## A.3 What happens during a call

### Tracing calls: each call gets its own variables

When a function is called, Python creates a fresh **frame**: a private workspace holding that call's parameters and variables. When the function returns, the frame disappears.

````python
def area(width, height):
    result = width * height
    return result

room = area(3, 4)
hall = area(2, 10)
print(room + hall)
````

Trace table for the two calls:

| Call | `width` | `height` | `result` | returns |
|---|---|---|---|---|
| `area(3, 4)` | 3 | 4 | 12 | 12 |
| `area(2, 10)` | 2 | 10 | 20 | 20 |

So `room` is 12, `hall` is 20, and the program prints **32**.

> 💡 To *see* frames appear and disappear, paste this code into [pythontutor.com](https://pythontutor.com) and step through it.

### Functions can call other functions

- `hypotenuse` uses `square` twice.
- Small functions combine into bigger ones.
- This is **abstraction**: you can use a function without thinking about what is inside it.

````python
import math

def square(x):
    return x * x

def hypotenuse(a, b):
    return math.sqrt(square(a) + square(b))

print(hypotenuse(3, 4))
````

````python
# ---- Practice 6A, Level 2: Apply ----
def rectangle_area(width, height):
    """Return the area of a rectangle."""
    return width * height

def rectangle_perimeter(width, height):
    """Return the perimeter of a rectangle."""
    return 2 * (width + height)

def count_vowels(text):
    """Return how many vowels (a, e, i, o, u) are in text, ignoring case."""
    count = 0
    for character in text.lower():
        if character in "aeiou":
            count += 1
    return count

print(rectangle_area(3, 4), rectangle_perimeter(3, 4))
print(count_vowels("Maiduguri"), count_vowels("RHYTHM"), count_vowels("ArewaDS"))
````

````python
# ---- Practice 6A, Level 3: Extend (Caesar cipher as two functions) ----
def shift_letter(character, shift):
    """Shift one letter forward by shift places, wrapping around the alphabet.
    Anything that is not a letter is returned unchanged."""
    if "a" <= character <= "z":
        base = ord("a")
    elif "A" <= character <= "Z":
        base = ord("A")
    else:
        return character
    return chr((ord(character) - base + shift) % 26 + base)

def caesar(text, shift):
    """Return text with every letter shifted by shift places."""
    result = ""
    for character in text:
        result += shift_letter(character, shift)
    return result

secret = caesar("Hello, World!", 3)
print(secret)
print(caesar(secret, -3))        # shifting back decodes it
print(caesar("xyz", 3))          # wraps around

assert caesar("abc", 3) == "def"
assert caesar("xyz", 3) == "abc"
assert caesar(caesar("ArewaDS", 5), -5) == "ArewaDS"
print("caesar: all tests passed")
````

## B.3 Functions that loop: reusable building blocks

### `is_prime`: return as soon as you know

A `return` inside a loop ends the **whole function**, not just the loop.

````python
def is_prime(n):
    """Return True if n is prime."""
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
    return True

print(is_prime(7), is_prime(9), is_prime(1))
````

Trace:

| Call | Divisors tried | Returns |
|---|---|---|
| `is_prime(7)` | 2, 3, 4, 5, 6 (none divide evenly) | `True` |
| `is_prime(9)` | 2, 3 (`9 % 3 == 0`, so stop) | `False` |
| `is_prime(1)` | none (`n < 2`) | `False` |

### ✍️ YOUR TURN: primes up to 30 (about 5 minutes)

Write `is_prime(n)` yourself (try not to scroll up!). Then use it in a loop to print every prime from 1 to 30 on **one line**.

Expected output:
```

```
**Hint:** `print(n, end=" ")` keeps output on one line.

````python
# Write is_prime(n) and the loop here
def is_prime(n):
    """Return True if n is prime, otherwise False."""
    if n < 2:                      # 0, 1 and negatives are not prime
        return False
    for divisor in range(2, n):    # try 2, 3, ... n - 1
        if n % divisor == 0:       # found a divisor, so n is not prime
            return False           # stop immediately
    return True                    # no divisor found: prime

for n in range(1, 31):
    if is_prime(n):
        print(n, end=" ")          # end=" " keeps everything on one line
print()                            # finish the line
````

### A reusable input helper

This is the 4B validation loop, written **once** as a function:

- `return` both exits the `while` loop and ends the function.
- Now any program can ask for a valid number in **one line**.

### ▶️ TRY IT
Run it and deliberately type bad input first: try `abc`, then `150`, then `19`.

````python
def get_int(prompt, low, high):
    """Ask until a whole number in low..high."""
    while True:
        try:
            value = int(input(prompt))
        except ValueError:
            print("Whole numbers only.")
            continue
        if low <= value <= high:
            return value
        print(f"Enter {low} to {high}.")

age = get_int("Age: ", 0, 120)
print("Age is", age)
````

## B.4 Refactoring: same behavior, better structure

**Refactoring** means changing how code is *organized* without changing what it *does*.

### Before: everything at the top level
Run it with 45 hours at a rate of 20.

````python
hours = float(input("Hours: "))
rate = float(input("Rate: "))
if hours <= 40:
    pay = hours * rate
else:
    extra = hours - 40
    pay = 40 * rate + extra * rate * 1.5
print(f"Pay: ${pay:.2f}")
````

### After: a function and `main()`
Run it with the same inputs.

````python
def weekly_pay(hours, rate):
    """Time and a half over 40 hours."""
    if hours <= 40:
        return hours * rate
    extra = hours - 40
    return 40 * rate + extra * rate * 1.5

def main():
    hours = float(input("Hours: "))
    rate = float(input("Rate: "))
    pay = weekly_pay(hours, rate)
    print(f"Pay: ${pay:.2f}")

main()
````

Both print `Pay: $950.00` for 45 hours at \$20. **Same behavior, better structure.**

### Now the logic can be tested on its own

- `weekly_pay` needs no `input()`, so it is easy to test.
- **Boundary:** exactly 40 hours.
- **Edge:** 0 hours.

````python
assert weekly_pay(40, 20) == 800
assert weekly_pay(45, 20) == 950
assert weekly_pay(0, 20) == 0
print("weekly_pay tests passed")
````

### Worked example: the grade report, assembled

Here is the full program planned in B.1. Notice:

- `main()` reads like the plan.
- Each helper was written and tested separately before being put together.

Run it. Try typing `ninety` for one score to see the validation work.

````python
def get_int(prompt, low, high):
    """Ask until a whole number in low..high."""
    while True:
        try:
            value = int(input(prompt))
        except ValueError:
            print("Whole numbers only.")
            continue
        if low <= value <= high:
            return value
        print(f"Enter {low} to {high}.")

def average(total, count):
    """Return total / count, or 0.0 if count is 0."""
    if count == 0:
        return 0.0
    return total / count

def letter_grade(score):
    """Return the letter for a 0-100 score."""
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    return "F"

def main():
    count = get_int("How many tests? ", 1, 10)
    total = 0
    for i in range(count):
        total += get_int(f"Score {i + 1}: ", 0, 100)
    avg = average(total, count)
    print(f"Average {avg:.1f}, grade {letter_grade(avg)}")

if __name__ == "__main__":
    main()
````

Score 2: ninety
Whole numbers only.

Score 3: 72
Average 85.0, grade B
